```python
from sklearn.model_selection import KFold
import json

def k_fold_split(data, k):
    """
    Splits the given dataset into K folds.
    :param data: List or numpy array of data.
    :param k: Number of folds.
    :return: List of tuples containing (train_indices, test_indices) for each fold.
    """
    kf = KFold(n_splits=k, shuffle=True, random_state=42)
    folds = [(train_index, test_index) for train_index, test_index in kf.split(data)]
    
    return folds

def get_fold_data(data, indices):
    """
    Retrieves the actual data based on provided indices.
    :param data: List or numpy array of data.
    :param indices: List of indices to retrieve.
    :return: List of data points corresponding to the indices.
    """
    return [data[i] for i in indices]


def save_to_jsonl(data, filename):
    """
    Save a list of dictionaries to a JSONL file.

    :param data: List of dictionaries to save.
    :param filename: The file path to save the JSONL file.
    """
    if not isinstance(data, list) or not all(isinstance(item, dict) for item in data):
        raise ValueError("Data must be a list of dictionaries")

    with open(filename, 'w', encoding='utf-8') as f:
        for item in data:
            f.write(json.dumps(item, ensure_ascii=False) + '\n')
```

```python
import jsonlines
file_name = "./alfworld/test.jsonl"
data_list = []
with open(file_name, 'r+', encoding='utf-8') as f:
    for item in jsonlines.Reader(f):
        data_list.append(item)

folds = k_fold_split(data_list, k=4)
# Display the split indices and retrieve actual data
for i, (train, test) in enumerate(folds):
    print(f"Fold {i+1}\n")
    train_data = get_fold_data(data_list, train)
    test_data = get_fold_data(data_list, test)
    save_to_jsonl(train_data, filename="./alfworld/train_data_%s.jsonl"%i)
    save_to_jsonl(test_data, filename="./alfworld/test_data_%s.jsonl"%i)
    
    # print('len(train_data)', len(train_data))
    # count_dict = dict()
    # for item in train_data:
    #     task_category = item["additional_info"]["description"].split('-')[0].strip()
    #     if task_category not in count_dict.keys():
    #         count_dict[task_category] = 1
    #     else:
    #         count_dict[task_category] += 1
    # for item in count_dict:
    #     print(item, count_dict[item])

    # print('len(test_data)', len(test_data))
    # count_dict = dict()
    # for item in test_data:
    #     task_category = item["additional_info"]["description"].split('-')[0].strip()
    #     if task_category not in count_dict.keys():
    #         count_dict[task_category] = 1
    #     else:
    #         count_dict[task_category] += 1
    # for item in count_dict:
    #     print(item, count_dict[item])
    # print('-------------------')


```
